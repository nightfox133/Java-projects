Below are the packages needed for this exercise. You do not need to modify the code block below.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

We generate clusters in the code block below.

For the first part of this exercise, we will create four non-overlapping clusters, each in distinct quadrants.

The centers of the first two clusters have already been set. Determine what the centers of Cluster 3 and 4 should be, given the conditions that each cluster is in its own quadrant and clusters are non-overlapping. There is a most obvious choice, but there is not necessarily a single correct answer.

Let's also say that each cluster has 75 to 125 points - you can choose the exact number yourself. Fill in the last parameter of the function call. This doesn't have to be the same for every cluster!

```python
#Cluster 1
center1 = [-5, 5]
cov1 = [[1, 0], [0, 1]]
cluster1 = np.random.multivariate_normal(center1, cov1, 75) #TODO

#Cluster 2
center2 = [-5, -5]
cov2 = [[2,0], [0,2]]
cluster2 = np.random.multivariate_normal(center2, cov2, 80) #TODO

#Cluster 3
center3 = [5, -5]
cov3 = [[3,0], [0,3]]
cluster3 = np.random.multivariate_normal(center3, cov3, 100) #TODO

#Cluster 4
center4 = [5, 5]
cov4 = [[0.5, 0], [0, 0.5]]
cluster4 = np.random.multivariate_normal(center4, cov4, 120) #TODO
```

This is code to visualize the cluster data. You do not need to make any edits to the code below.

```python
plt.figure(figsize=(7, 7))

plt.scatter(cluster1[:, 0], cluster1[:, 1], c = "blue")
plt.scatter(cluster2[:, 0], cluster2[:, 1], c = "blue")
plt.scatter(cluster3[:, 0], cluster3[:, 1], c = "blue")
plt.scatter(cluster4[:, 0], cluster4[:, 1], c = "blue")

plt.title('Observations with 4 Distinct Clusters')
plt.xlabel(r'$x_1$')
plt.ylabel(r'$x_2$')
plt.show()
```

Below is the code to apply K-means clustering. You will run this code at least twice:
1. Once with initial centers of (-4, 0), (5,8), (3,4), (7,0)
2. Try with different initial centers until you find a combination that produces the "ideal" clustering

Fill in the parts that say TODO

```python
#K-MEANS CLUSTERING

#combine data
all_data = np.concatenate((cluster1, cluster2, cluster3, cluster4))

#set initial centers
init_centers = np.array([(-4, 0), (5,8), (3,4), (7,0)]) #TODO

#modeling
kmeans = KMeans(init=init_centers, n_clusters=4, n_init=1, random_state=42)
kmeans.fit(all_data)
labels = kmeans.labels_

#plotting: you do not need to change anything below
#----------------------------------------------------
plt.figure(figsize=(5, 5))
colors = ['blue', 'red', 'green', 'purple']
symbols = ['o', 's', '^', '+']

for i in range(kmeans.n_clusters):
    cluster_data = all_data[labels == i] #get data for specific cluster
    plt.scatter(cluster_data[:, 0], cluster_data[:, 1], c = colors[i], marker = symbols[i])

plt.title("K-Means on 4 Distinct Clusters")
plt.xlabel(r'$x_1$')
plt.ylabel(r'$x_2$')
plt.show()
```
